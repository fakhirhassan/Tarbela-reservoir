```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# --- Toy twin simulation ---
def simulate_twin(state="normal", noise=0.0):
    t = np.linspace(0, 2, 2000)
    if state=="normal":
        sig = np.sin(2*np.pi*50*t) * (np.sin(2*np.pi*2*t)>0)
    else:  # murmur
        sig = np.sin(2*np.pi*50*t) * (np.sin(2*np.pi*2*t)>0) + 0.3*np.random.randn(len(t))
    sig = sig + noise*np.random.randn(len(t))
    return t, sig

# --- Interactive dashboard ---
def plot_twin(state="normal", noise=0.0):
    t, sig = simulate_twin(state, noise)
    plt.figure(figsize=(10,3))
    plt.plot(t, sig)
    plt.title(f"Digital Twin Simulation: {state}, noise={noise}")
    plt.xlabel("Time [s]")
    plt.ylabel("Amplitude")
    plt.show()

widgets.interact(plot_twin,
                 state=["normal","murmur"],
                 noise=widgets.FloatSlider(min=0,max=1,step=0.1,value=0.0))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import glob, os
import ipywidgets as widgets

# --- Collect some real PCG files ---
normal_files = glob.glob("../data/train/normal/*.wav")
murmur_files = glob.glob("../data/train/murmur/*.wav")

# Global holder for last signal
last_signal = {"t": None, "sig": None, "sr": None, "fname": None}

def simulate_twin(state="Normal", noise=0.0, sr=2000):
    global last_signal
    if state == "Normal":
        file = np.random.choice(normal_files)   # pick random normal
    else:
        file = np.random.choice(murmur_files)   # pick random murmur
    
    # Load original PCG
    sig, sr = librosa.load(file, sr=sr)
    
    # Add noise if requested
    if noise > 0:
        sig = sig + noise * np.random.randn(len(sig))
    
    # Time vector
    t = np.arange(len(sig)) / sr
    
    # Save globally
    last_signal = {"t": t, "sig": sig, "sr": sr, "fname": os.path.basename(file)}
    
    return t, sig, os.path.basename(file)

```

```python
# Function to update heart based on state
def update_heart(state="Normal", noise=0.0):
    # Map state to color and text
    global status_text 

    status_text = state
```

```python

    
# --- Extended Plot function ---
def plot_twin(state="Normal", noise=0.0):
    t, sig, fname = simulate_twin(state, noise)
    update_heart(state, noise)
    # Sampling rate from twin simulation
    sr = int(1 / (t[1] - t[0]))
    
    plt.figure(figsize=(15, 8))
    
    # 1) Waveform
    plt.subplot(3, 1, 1)
    plt.plot(t, sig, color="blue")
    plt.title(f"Waveform - Digital Twin ({state}), noise={noise}, file={fname}")
    plt.xlabel("Time [s]")
    plt.ylabel("Amplitude")
    
    # 2) FFT / Frequency Spectrum
    plt.subplot(3, 1, 2)
    freqs = np.fft.rfftfreq(len(sig), d=1/sr)
    fft_vals = np.abs(np.fft.rfft(sig))
    plt.plot(freqs, fft_vals, color="purple")
    plt.title("Frequency Spectrum (FFT)")
    plt.xlabel("Frequency [Hz]")
    plt.ylabel("Magnitude")
    plt.xlim(0, 500)  # heart sounds mostly <500 Hz
    
    

# --- Interactive dashboard ---
widgets.interact(
    plot_twin,
    state=["Normal", "Abnormal"],
    noise=widgets.FloatSlider(min=0, max=1, step=0.1, value=0.0)
)
```
