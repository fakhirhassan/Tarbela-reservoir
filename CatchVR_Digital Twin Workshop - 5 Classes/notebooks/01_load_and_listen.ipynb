```python
import os
import librosa
import matplotlib.pyplot as plt
import IPython.display as ipd

# --- Paths ---
data_dir = "../data/train/AS"   #directory name from where we want to read data like ../data/train/normal   # change to murmur if needed
files = [f for f in os.listdir(data_dir) if f.endswith(".wav")]
print("Available files:", files[:5])

# --- Load one file with fixed sr=2000 ---
file = os.path.join(data_dir, files[0])
x, sr = librosa.load(file, sr=2000)   # resamples if needed
print(f"Loaded: {file} | Shape: {x.shape} | Sample Rate: {sr}")

# --- Listen ---
ipd.Audio(x, rate=sr)

# --- Plot ---
plt.figure(figsize=(12,3))
plt.plot(x)
plt.title("Raw PCG waveform (sr=2000)")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.show()
```
